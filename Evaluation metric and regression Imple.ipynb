```python
#Q1.  What does R-squared represent in a regression model?
"""
Ans1. R-square is also known as Coefficient of dtermination, measures how well the regression model
       explains the variation in the dependent variable.
       The percentage variation of in the dependent variable (y) that can be explained 
       by independent variabel (x) is called R square.

       For example: an R-squared value of 0.75 means 75% of the variance in y is explained by the model,
                    while the remaining 25% is due to other factors or noise.
"""
```

```python
#Q2.  What are the assumptions of linear regression?
"""
Ans2. Assumptions of linear regression.
        1. Linearity:
                Independent varibale (X) and dependent variable(y) should have linear relationship.
        2. Independence:
                The observation (rows) are independent of each other.
        3. Homoscedasticity:
                Also knows as constant variance, this assumption means the variance of the residuals(error)
                are constant across all levels of the independent varibales (means no pattern).
        4. Normality of error:
                Residuals (Error) should follow a normal distribution.
        5. Multicollinearity:
                Independent variables should not be highly correlated with each other.
"""
```

```python
#Q3.  What is the difference between R-squared and Adjusted R-squared?
"""
Ans3. - R-square shows how much of the variation in the dependent variable is explained by the model.
        Adding more variables will always make it go up.
      - Adjusted R square does the same but also checks if the new varibale are actually useful. If a
        variables doesn't help, Adjusted R-squared may go down.
        To undersatnt if added feature/I.V is contributing to the performance or not,
        we have adjusted r square.
"""
```

```python
#Q4. Why do we use Mean Squared Error (MSE)?
"""
Ans4. Mean squared error measures the average of squared differences between the predicted and actual
      value in regression.
      Why we use MSE.
      1. No cancellation of error:
                  Squaring ensures positive values, so overprediction and underprediction don't 
                  cancel out each other.
      2. Cost Function:
                  MSE is used as cost function.
      3. Differentiable:
                  MSE is easy to differentiate, making it suitable for optimization algorithms like
                  gradient descent.
      4. Convex Function:
                  It has one one global minimum, which makes optimization straightforward.
      5. Emphasis on large error:
                  Squaring amplified the impact of large error.
"""
```

```python
#Q5.  What does an Adjusted R-squared value of 0.85 indicate?
"""
Ans5. An adjusted R-squared value of 0.85 indicates that approximately 85% of the variation in the 
      dependent variable is explained by the independent variables in your regression model, after
      accounting for the number of predictors.
      This means the model fits of the data well, and the independent variables collectively provide a
      strong explanation of the dependent varibale's behaviour.
"""
```

```python
#Q6.  How do we check for normality of residuals in linear regression?
"""
Ans6. To check the normality of residuals in linear regression, there are some methods are:
        1. Histogram:
                Plot a histogram of the residuals ans visually inspect if they follow a bell shaped
                curve.
        2. Q-Q Plot (Quantile Quantile Plot):
                Compare the quantiles of the residuals to a normal distribution. If the points lie 
                close to the diagonal line, the residuals are likely normally distributed.
        3. Shapiro-walk test:
                A statstical test that checks if the residuals are normally distributed. A p-value
                greater than 0.5 indicates normality.
        4. Kolmogorov-Smirnov test:
                Another statistical test for normality, comparing the residual to a normal
                distribution.
        5. Skewness and Kurtosis:
                calculate these metrics to asses the shape of the residuals distribution.
                value close to zero suggest normality.
"""
```

```python
#Q7. What is multicollinearity, and how does it impact regression?
"""
Ans7. Multicollinearity means relationship among multiple features.
      Multicollinarity occurs when two or more independent varibales in a regression model are highly
      correlated.
      This makes it difficult for the model to distinguish their individual effects on dependent variable
      on the dependent variable, as they share similar information.

      How does it affect regression:
      1. Affects Interpreattion:
              Its hard to know which feature is contributing to what extent in change of (dependent variable)y.
      2. Increases Computation:
              If x1 = x2, theny why to use both the feature in buliding the model. It will increase computation.
      3. It increase Overfitting:
              Both x1 and x2 is almost same the pattern will be the memorised that's why it leads to overfitting.
"""
```

```python
#Q12. What does a high R-squared value mean?
"""
Ans12. A high R-squared value indicates that the independent variables in the regression model
       explain a large proportion of the variance in the dependent variable.
       For example, if R square = 0.90, it means 90% of the variation in the dependent variable
                    is accounted for by the mode, while the remaining 10% is due to other factors
                    or randomness.
        - A High R squared doesn't always mean the model is good; the model could be overfitting
          the data.
        - For certain types of data(e.g., non- linear relationships), R-squared might not be
          a sufficient measure for model performance.
"""
```

```python
#Q13.  What happens if linear regression assumptions are violated?
"""
Ans13. If linear regression assumptions are violated, the model reliability and accuracy can be
       compromised. 
       1. Linearity Violation:
               If the relationship between variables isn't linear, the model may fail to capture the 
               true pattern, leading to biased predictions.
       2. Independence Violation:
               When observation are not independent, standard errors can be underestimated, affected
               hypthesis testing.
       3. Homoscedasticity Violation:
               If residuals have non-constant variance(hetroscedasticity), the model's predictions
               become less reliable, and statistical tests lose power.
       4. Normality Violation:
               Non-normal residuals can impact the validity  of confidence inervals and hypothesis 
               tests.
"""
```

```python
#Q14.  How can we address multicollinearity in regression?
"""
Ans14. To adress multicollinearity in regression, use following startegies:

        1. VIF (Variation Inflation Factor):
                Calculate VIF for each variable, If a variable has a high VIF(e.g., > 10),
                consider removing it one by one or combing it with others.
        2. PCA (Principal Component Analysis):
                Transform the correlted variables into a smaller set of uncorrelated components.
        3. Regularization Techniques:
                Use methods like Ridge Regression or Lasso Regression, which can handle multicollinearity
                by penalizing the large coefficients.
        4. Combine Variables:
                If two variabels are highlt correlated, you can combine into a single feature.
        5. Collect More data:
                Increasing the sample size can sometimes reduce the impact of multicollinearity.
"""
```

```python
#Q17.  Why is MSE sensitive to outliers?
"""
Ans17. Mean Squared Error (MSE) is a sensitive to outliers because it squares the errors (difference
       between actual and predicted values). Squaring makes larger errors even bigger, so outliers
       heavily influence the final MSE value.
       This makes the metric less reliable if your data contains extreme values.
"""
```

```python
#Q18. What is the role of homoscedasticity in linear regression?
"""
Ans18. Homoscedasticity playes a crucial role in linear regression as it ensures the reliability
       of the model' prediction and statistical tests.
       It is important because:

       1. Constant Variance:
               Homoscedasticity means the residuals (errors) have a constant variance across all levels
               of the independent variabkes. The consistency is essential for acctual predictions.
       2. Valid Statistical Tests:
               Many Statistical Tests, like T-tests and F- tests, assume homoscedasticity. Violation
               can lead to incorrect conclusions about the significance of predictors.
       3. Reliable Confidence Intervals:
               Homoscedasticity ensures that confidence intervals for coefficients are accurate, 
               supporting better decision making.
       4. Model Stability:
               It helps maintain the stability of the regression model, preventing biased estimates.
"""
```

```python
#Q19. What is Root Mean Squared Error (RMSE)?
"""
Ans19. It is calculated as the square root of the Mean Squared Error(MSE), which makes it easier
       to interpret because RMSE is expressed in the same units as the dependent variable.
       
"""
```

```python
#Q22.  What is heteroscedasticity, and why is it a problem?
"""
Ans22. Hetroscedasticity occurs when the variance of residuals in a regression model is not constant
       across all levels of the independent variables.
       Why it is problem:
       1. Reduced Efficiency:
               The model's estimates become less precise, affecting its overall reliability.
       2. Biased Predictions:
               Unequal variance can disort the model's predictions, especially for extreme values.
       3. Invalid Statistical Tets:
               Many regression techniques assume constant variance(homoscedasiticity). Violation can
               lead to unreliable hypothesis tests and confidence intervals.
       
"""
```

```python
#Q11. Write a Python script that fits polynomial regression models of different degrees to a synthetic dataset and 
#     compares their performance.

import numpy as np
import pandas as pd
import seaborn as sns
```

```python
#Generate Synthetic dataset
np.random.seed(1)
X = 2 * np.random.rand(100,1)
y = 4 + 3*X+1.5*X**2 + np.random.randn(100,1)
```

```python
#split the data into training and tesing
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state= 1)
```

```python
degrees = [1,2,3,4,5]
mse_scores = []
r2_scores = []
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
for degree in degrees:
    #transform the features for the given degree
    poly = PolynomialFeatures(degree = degree)
    X_poly_train = poly.fit_transform(X_train)
    X_poly_test = poly.transform(X_test)

    #train the model
    model = LinearRegression()
    model.fit(X_poly_train,y_train)

    #predict the model
    y_pred = model.predict(X_poly_test)

    #Evaluate the model
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    mse_scores.append(mse)
    r2_scores.append(r2)
```

```python
#display peformance metrics
for i, degree in enumerate(degrees):
    print(f"Degree {degree}: MSE = {mse_scores[i]:.2f}, R^2 = {r2_scores[i]:.2f}")
```

```python
#Q23. Compare Ridge Regression with Linear Regression on a synthetic dataset and print the coefficients and R
#    squared score
```

```python
np.random.seed(1)
X = 2 * np.random.rand(100,1)
y = 4 + 3*X+1.5**2 + np.random.randn(100,1)
```

```python
#split the data into training set and test set
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2, random_state=1)
```

```python
from sklearn.linear_model import LinearRegression,Ridge
```

```python
#LinearRegression
model_lin = LinearRegression()
model_lin.fit(X_train,y_train)
y_pred_lin = model_lin.predict(X_test)
```

```python
#Ridge Regression
model_reg = Ridge(alpha = 1.0)
model_reg.fit(X_train,y_train)
y_pred_ridge = model_reg.predict(X_test)
```

```python
from sklearn.metrics import r2_score
```

```python
print("Linear Regression Coefficient", model_lin.coef_)
print("Linear Regression Intercept", model_lin.intercept_)
print("Linear Regression R2 score", r2_score(y_test,y_pred_lin))


print("Ridge Regression Coefficient", model_reg.coef_)
print("Ridge Regression Intercept", model_reg.intercept_)
print("Ridge Regression R2 score", r2_score(y_test,y_pred_ridge))
```

```python
#Q25. Write a Python script that compares polynomial regression models of different degrees and prints the R
#     squared score for each
```

```python
#Generate Synthetic dataset
np.random.seed(1)
X = 2 * np.random.rand(100,1)
y = 4 + 3*X+1.5*X**2 + np.random.randn(100,1)
```

```python
#split the data into training and tesing
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state= 1)
```

```python
degrees = [1,2,3,4,5]
r2_scores = []
```

```python
for degree in degrees:
    #transform the features for the given degree
    poly = PolynomialFeatures(degree = degree)
    X_poly_train = poly.fit_transform(X_train)
    X_poly_test = poly.transform(X_test)

    #train the model
    model = LinearRegression()
    model.fit(X_poly_train,y_train)

    #predict the model
    y_pred = model.predict(X_poly_test)

    #Evaluate the model
    r2 = r2_score(y_test, y_pred)
    r2_scores.append(r2)
```

```python
#display peformance metrics
for i, degree in enumerate(degrees):
    print(f"R^2 = {r2_scores[i]:.2f}")
```
